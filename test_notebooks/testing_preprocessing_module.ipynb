```python
from pathlib import Path
from typing import Callable, Optional

import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset
import zarr


class ZarrEEGRecordingDataset(Dataset):
    """
    Lazily loads exam-level EEG window arrays from the chunked zarr layout
    written by `save_site_in_chunks`:

        root/
            site-i/chunk_0000/data.zarr      shape (n_exams, n_windows, n_channels, n_samples)
            site-i/chunk_0000/metadata.parquet
            ...

    Each chunk stores one exam per zarr chunk (chunks=(1, n_windows, ...)),
    so a single-index read on axis 0 touches exactly one on-disk chunk —
    __getitem__ never reads more than one exam's worth of data.

    Each item is (x, y):
        x : FloatTensor, shape (n_windows, n_channels, n_samples)
        y : LongTensor scalar, the exam's target
    """

    def __init__(self,
                root,
                target_col: str = "target",
                transform: Optional[Callable[[np.ndarray], np.ndarray]] = None):
        self.root = Path(root)
        self.target_col = target_col
        self.transform = transform

        # (zarr_array, local_idx, metadata_row) triples, one per exam, built in
        # a single pass so data and labels can never drift apart after
        # construction — unlike indexing into a separately-concatenated
        # metadata DataFrame by a global position that happens to line up.
        self._items: list[tuple[zarr.Array, int, pd.Series]] = []

        expected_shape = None  # (n_windows, n_channels, n_samples), validated across chunks

        for site_dir in sorted(self.root.iterdir()):
            if not site_dir.is_dir():
                continue

            for chunk_dir in sorted(site_dir.glob("chunk_*")):
                zarr_path = chunk_dir / "data.zarr"
                metadata_path = chunk_dir / "metadata.parquet"

                if not zarr_path.exists() or not metadata_path.exists():
                    continue

                data = zarr.open(zarr_path, mode="r")["data"]
                metadata = pd.read_parquet(metadata_path)

                n_exams = data.shape[0]
                if n_exams != len(metadata):
                    raise ValueError(
                        f"{chunk_dir}: {n_exams} exams in data.zarr but "
                        f"{len(metadata)} rows in metadata.parquet"
                    )

                chunk_shape = data.shape[1:]  # (n_windows, n_channels, n_samples)
                if expected_shape is None:
                    expected_shape = chunk_shape
                elif chunk_shape != expected_shape:
                    raise ValueError(
                        f"{chunk_dir}: window shape {chunk_shape} does not match "
                        f"{expected_shape} seen in earlier chunks"
                    )

                for local_idx in range(n_exams):
                    self._items.append((data, local_idx, metadata.iloc[local_idx]))

        if not self._items:
            raise RuntimeError(f"No usable (data.zarr, metadata.parquet) pairs found under {self.root}")

    def __len__(self) -> int:
        return len(self._items)

    def get_metadata(self, idx: int) -> pd.Series:
        """Metadata row for exam `idx`, without loading its array data."""
        return self._items[idx][2]

    def __getitem__(self, idx: int):
        data, local_idx, row = self._items[idx]

        # Single-index read: touches exactly one on-disk zarr chunk.
        x = data[local_idx]
        if self.transform is not None:
            x = self.transform(x)
        x = torch.from_numpy(np.asarray(x, dtype=np.float32))

        y = torch.tensor(row[self.target_col], dtype=torch.long)

        return x, y
```
